```python
# Cell 3: Kör Alt 2397 Block 3 (276-280, 10+250 år, 3 armar = 15 körningar) med live-utskrift och nedladdning
import os, subprocess, sys
from google.colab import files
nproc = os.cpu_count() or 8
ra_ut = '/content/alt2397-b3.json'
md_ut = '/content/alt2397-b3.md'
print(f'Startar Alt 2397 Block 3 (276-280) på {nproc} CPU-kärnor...', flush=True)
env = dict(os.environ, HCSIM_ROT='/content/hcsim-colab/n10/hcsim')
cmd = [
    '/content/hcsim-colab/n10/hcsim/target/release/prov2008',
    'sparb2397', '--fron', '276-280', '--steg', '1',
    '--inkor', '10', '--ar', '250', '--tradar', str(nproc),
    '--ra-ut', ra_ut, '--ut', md_ut
]
proc = subprocess.Popen(cmd, env=env, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
for line in proc.stdout:
    sys.stdout.write(line)
    sys.stdout.flush()
rc = proc.wait()
assert rc == 0, f'prov2008 avslutade med {rc}'
files.download(ra_ut)
files.download(md_ut)
```
